# 7.1. Data Cleaning

In this section, we will clean the original Esports Earning History dataset using Python and Pandas.

The cleaning process includes:

- Standardizing column names.
- Removing unnecessary spaces.
- Correcting data types.
- Identifying missing and inconsistent values.
- Checking invalid numerical values.
- Validating records and identifiers.

The original datasets will remain unchanged. All cleaned files will be saved in the processed folder.

```python
# Importar bibliotecas
import pandas as pd
import numpy as np
import os

# Definir rutas
RAW_PATH = "../data/raw/"
PROCESSED_PATH = "../data/processed/"

# Crear carpeta de salida si no existe
os.makedirs(PROCESSED_PATH, exist_ok=True)

# Cargar los archivos originales
# keep_default_na=False evita que "NA" sea interpretado como nulo.
# na_values=[""] mantiene los campos vacíos como valores faltantes.

country = pd.read_csv(
    RAW_PATH + "esports_country_earnings_by_year.csv.csv",
    keep_default_na=False,
    na_values=[""]
)

game = pd.read_csv(
    RAW_PATH + "esports_game_earnings_by_year.csv.csv",
    keep_default_na=False,
    na_values=[""]
)

player = pd.read_csv(
    RAW_PATH + "esports_player_earnings_by_year.csv.csv",
    keep_default_na=False,
    na_values=[""]
)

team_game = pd.read_csv(
    RAW_PATH + "esports_team_earnings_by_game.csv.csv",
    keep_default_na=False,
    na_values=[""]
)

team_year = pd.read_csv(
    RAW_PATH + "esports_team_earnings_by_year.csv.csv",
    keep_default_na=False,
    na_values=[""]
)

print("Los cinco archivos se cargaron correctamente.")
```

```python
# Lista de tablas
datasets = {
    "country": country,
    "game": game,
    "player": player,
    "team_game": team_game,
    "team_year": team_year
}

# Estandarizar nombres de columnas
for name, df in datasets.items():

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
    )

    # Eliminar espacios al principio y al final
    # de las columnas de texto
    for column in df.select_dtypes(include=["str", "object"]).columns:
        df[column] = df[column].str.strip()

print("Nombres de columnas estandarizados.")
```

```python
# Recuperar el código ISO de Namibia
country.loc[
    (country["country_name"] == "Namibia") &
    (country["country_code"].isna()),
    "country_code"
] = "NA"

print("Códigos de país corregidos:")
print(
    country[country["country_name"] == "Namibia"][
        ["year", "country_name", "country_code"]
    ]
)
```

```python
# Recuperar el código ISO de Namibia
country.loc[
    (country["country_name"] == "Namibia") &
    (country["country_code"].isna()),
    "country_code"
] = "NA"

print("Códigos de país corregidos:")
print(
    country[country["country_name"] == "Namibia"][
        ["year", "country_name", "country_code"]
    ]
)
```

```python
# Contar nombres desconocidos
unknown_names = (
    player["player_name"].eq("- -").sum()
)

print("Nombres desconocidos encontrados:", unknown_names)

# Convertir el marcador - - en un valor faltante
player["player_name"] = player["player_name"].replace(
    "- -",
    pd.NA
)

print(
    "Nombres desconocidos después de la limpieza:",
    player["player_name"].isna().sum()
)
```

```python
# Columnas numéricas por tabla
numeric_columns = {
    "country": [
        "year", "rank", "total_earnings",
        "number_of_players"
    ],

    "game": [
        "year", "rank", "total_earnings",
        "number_of_players", "number_of_tournaments"
    ],

    "player": [
        "year", "rank", "player_website_id",
        "total_year", "total_overall",
        "percentage_of_total", "age"
    ],

    "team_game": [
        "team_website_id", "game_rank_for_team",
        "team_total_earnings", "overall_total_earnings",
        "percentage_of_overall"
    ],

    "team_year": [
        "year", "rank", "team_website_id",
        "total_earnings", "number_of_tournaments"
    ]
}

# Convertir columnas numéricas
for name, df in datasets.items():

    for column in numeric_columns[name]:

        df[column] = pd.to_numeric(
            df[column],
            errors="raise"
        )

    # Convertir años e identificadores a enteros anulables
    integer_columns = [
        col for col in numeric_columns[name]
        if col in ["year", "rank", "player_website_id",
                   "team_website_id", "game_rank_for_team",
                   "number_of_players", "number_of_tournaments"]
    ]

    for column in integer_columns:
        df[column] = df[column].astype("Int64")

print("Tipos de datos revisados y corregidos.")
```

```python
# Revisar valores numéricos negativos

for name, df in datasets.items():

    numeric_df = df.select_dtypes(include="number")

    print("\nTabla:", name)

    negative_values = (
        numeric_df.select_dtypes(exclude="bool") < 0
    ).sum()

    print(
        negative_values[negative_values > 0]
    )

    if not (numeric_df.select_dtypes(exclude="bool") < 0).any().any():
        print("No se encontraron valores negativos.")
```

```python
# Validar porcentajes entre 0 y 100

print(
    "Porcentajes de jugadores fuera del rango:",
    (
        (player["percentage_of_total"] < 0) |
        (player["percentage_of_total"] > 100)
    ).sum()
)

print(
    "Porcentajes de equipos fuera del rango:",
    (
        (team_game["percentage_of_overall"] < 0) |
        (team_game["percentage_of_overall"] > 100)
    ).sum()
)
```

```python
# Revisar duplicados completos

for name, df in datasets.items():

    duplicates = df.duplicated().sum()

    print(
        f"{name}: {duplicates} registros duplicados"
    )

    # Eliminar únicamente filas completamente duplicadas
    datasets[name] = df.drop_duplicates().copy()

# Actualizar las variables
country = datasets["country"]
game = datasets["game"]
player = datasets["player"]
team_game = datasets["team_game"]
team_year = datasets["team_year"]
```

# 7.2. Missing Values Treatment

In this section, we will identify and handle missing values using Pandas.

Different strategies will be applied depending on the type of information and its importance for the analysis.

The objective is to preserve useful records without introducing incorrect or invented information.

```python
# Revisar valores faltantes en las tablas

for name, df in datasets.items():

    print("\nDataset:", name)

    missing = df.isnull().sum()

    print(missing[missing > 0])

    print(
        "Total missing values:",
        df.isnull().sum().sum()
    )
```

```python
# Mantener las edades desconocidas como valores nulos

player["age"] = player["age"].astype("float64")

print("Edades faltantes:", player["age"].isna().sum())
```

```python
# Verificar nombres faltantes

print(
    "Nombres de jugadores faltantes:",
    player["player_name"].isna().sum()
)
```

```python
# Generar reporte de valores faltantes

missing_report = []

for name, df in datasets.items():

    for column in df.columns:

        missing_count = df[column].isna().sum()

        if missing_count > 0:

            missing_report.append({
                "Dataset": name,
                "Column": column,
                "Missing Values": missing_count,
                "Percentage": round(
                    missing_count / len(df) * 100, 2
                )
            })

missing_df = pd.DataFrame(missing_report)

display(missing_df)
```

# 7.3. Data Model Structuring

The Esports Earning History dataset contains five tables with different levels of detail.

To organize the information correctly, we will use a dimensional data model with separate fact tables and shared dimensions.

This structure will allow us to analyze esports earnings by year, country, game, player, and team without duplicating financial values.

```python
# Dimensión de años

years = []

for df in [country, game, player, team_year]:

    if "year" in df.columns:
        years.extend(df["year"].dropna().tolist())

dim_year = pd.DataFrame({
    "year": sorted(set(years))
})

display(dim_year.head())
```

```python
# Dimensión de países

dim_country = pd.concat([
    country[["country_code", "country_name"]],
    player[["country_code", "country_name"]]
]).drop_duplicates()

dim_country = dim_country.sort_values(
    "country_name"
).reset_index(drop=True)

display(dim_country.head())
```

```python
# Dimensión de videojuegos

dim_game = pd.concat([
    game[["game_name", "genre"]],
    team_game[["game_name"]].assign(genre=pd.NA),
    player["games"].dropna().str.split(", ")
    .explode().to_frame(name="game_name")
    .assign(genre=pd.NA)
], ignore_index=True)

dim_game["game_name"] = dim_game["game_name"].str.strip()

dim_game = dim_game.dropna(
    subset=["game_name"]
).drop_duplicates(
    subset=["game_name"]
).sort_values(
    "game_name"
).reset_index(drop=True)

display(dim_game.head())
```

```python
# Dimensión de jugadores

dim_player = player[
    ["player_website_id", "player_id", "player_name"]
].drop_duplicates(
    subset=["player_website_id"]
).reset_index(drop=True)

display(dim_player.head())
```

```python
# Dimensión de equipos

dim_team = pd.concat([
    team_year[["team_website_id", "team_name"]],
    team_game[["team_website_id", "team_name"]]
]).drop_duplicates(
    subset=["team_website_id"]
).reset_index(drop=True)

display(dim_team.head())
```

```python
# Validar claves de las tablas

keys = {
    "country": ["year", "country_name"],
    "game": ["year", "game_name"],
    "player": ["year", "player_website_id"],
    "team_year": ["year", "team_website_id"],
    "team_game": ["team_website_id", "game_name"]
}

for name, columns in keys.items():

    duplicates = datasets[name].duplicated(
        subset=columns
    ).sum()

    print(f"\nTabla: {name}")
    print("Clave:", columns)
    print("Registros repetidos:", duplicates)

    if duplicates == 0:
        print("Validación correcta.")
    else:
        print("Se requiere revisar la clave.")
```

```python
# Guardar las tablas procesadas

cleaned_datasets = {
    "esports_country_earnings_by_year_clean.csv": country,
    "esports_game_earnings_by_year_clean.csv": game,
    "esports_player_earnings_by_year_clean.csv": player,
    "esports_team_earnings_by_game_clean.csv": team_game,
    "esports_team_earnings_by_year_clean.csv": team_year
}

for filename, df in cleaned_datasets.items():

    output_path = os.path.join(
        PROCESSED_PATH,
        filename
    )

    df.to_csv(
        output_path,
        index=False,
        encoding="utf-8-sig"
    )

    print("Archivo guardado:", filename)
```

```python
# Exportar dimensiones

dimensions = {
    "dim_year.csv": dim_year,
    "dim_country.csv": dim_country,
    "dim_game.csv": dim_game,
    "dim_player.csv": dim_player,
    "dim_team.csv": dim_team
}

for filename, df in dimensions.items():

    output_path = os.path.join(
        PROCESSED_PATH,
        filename
    )

    df.to_csv(
        output_path,
        index=False,
        encoding="utf-8-sig"
    )

    print("Dimensión guardada:", filename)
```

```python
# Mostrar archivos finales

processed_files = os.listdir(PROCESSED_PATH)

print("ARCHIVOS PROCESADOS:")
print("----------------------------")

for file in processed_files:
    print(file)

print("----------------------------")
print("Total de archivos:", len(processed_files))
```
