```python
import pandas as pd
import numpy as np
import os
import glob

print("Bibliotecas cargadas correctamente")
```

```python
# Ruta donde se encuentran los archivos originales
DATA_PATH = "../data/raw/"

# Buscar todos los archivos CSV
csv_files = glob.glob(os.path.join(DATA_PATH, "*.csv"))

# Mostrar los archivos encontrados
print("ARCHIVOS CSV ENCONTRADOS:")
print("--------------------------------")

for file in csv_files:
    print(os.path.basename(file))

print("--------------------------------")
print("Total de archivos:", len(csv_files))
```

```python
# Revisar las dimensiones de cada archivo CSV

for file in csv_files:

    df = pd.read_csv(file)

    print("\nArchivo:", os.path.basename(file))
    print("Filas:", df.shape[0])
    print("Columnas:", df.shape[1])
```

```python
# Examinar la estructura de los archivos

for file in csv_files:

    df = pd.read_csv(file)

    print("\n" + "=" * 60)
    print("ARCHIVO:", os.path.basename(file))
    print("=" * 60)

    print("\nColumnas:")
    print(df.columns.tolist())

    print("\nTipos de datos:")
    print(df.dtypes)

    print("\nPrimeros cinco registros:")
    print(df.head())
```

```python
# Revisar valores faltantes en cada archivo

for file in csv_files:

    df = pd.read_csv(file)

    print("\nArchivo:", os.path.basename(file))

    missing = df.isnull().sum()

    print(missing[missing > 0])

    print(
        "Total de valores faltantes:",
        df.isnull().sum().sum()
    )
```

```python
# Revisar registros duplicados

for file in csv_files:

    df = pd.read_csv(file)

    print("\nArchivo:", os.path.basename(file))

    print(
        "Registros duplicados:",
        df.duplicated().sum()
    )
```

```python
# Generar inventario general de los archivos

inventory = []

for file in csv_files:

    df = pd.read_csv(file)

    inventory.append({
        "Archivo": os.path.basename(file),
        "Filas": df.shape[0],
        "Columnas": df.shape[1],
        "Valores faltantes": df.isnull().sum().sum(),
        "Duplicados": df.duplicated().sum(),
        "Memoria (MB)": round(
            df.memory_usage(deep=True).sum() / (1024 ** 2), 2
        )
    })

inventory_df = pd.DataFrame(inventory)

display(inventory_df)
```
